```python
from abc import ABC, abstractmethod
class KamarHotel(ABC):
    def __init__(self, nama_kamar, stok, harga_dasar):
        self.nama_kamar = nama_kamar
        self.__stok = stok
        self.__harga_dasar = harga_dasar

    def get_stok(self):
        return self.__stok

    def tambah_stok(self, jumlah):
        if jumlah < 0:
            print(f"Gagal update stok {self.nama_kamar}! Stok tidak boleh negatif ({jumlah}).")
        else:
            self.__stok += jumlah
            print(f"Berhasil menambahkan stok {self.nama_kamar}: {jumlah} unit.")

    def get_harga_dasar(self):
        return self.__harga_dasar

    @abstractmethod
    def tampilkan_detail(self):
        pass

    @abstractmethod
    def hitung_harga_total(self, jumlah_malam):
        pass


class KamarDeluxe(KamarHotel):
    def __init__(self, nama_kamar, stok, harga_dasar, fasilitas):
        super().__init__(nama_kamar, stok, harga_dasar)
        self.fasilitas = fasilitas

    def tampilkan_detail(self):
        print(f"[DELUXE] {self.nama_kamar} | Fasilitas: {self.fasilitas}")

    def hitung_harga_total(self, jumlah_malam):
        pajak = self.get_harga_dasar() * 10 / 100
        return (self.get_harga_dasar() + pajak) * jumlah_malam


class KamarStandard(KamarHotel):
    def __init__(self, nama_kamar, stok, harga_dasar, kapasitas):
        super().__init__(nama_kamar, stok, harga_dasar)
        self.kapasitas = kapasitas

    def tampilkan_detail(self):
        print(f"[STANDARD] {self.nama_kamar} | Kapasitas: {self.kapasitas}")

    def hitung_harga_total(self, jumlah_malam):
        pajak = self.get_harga_dasar() * 5 / 100
        return (self.get_harga_dasar() + pajak) * jumlah_malam


def proses_transaksi(daftar_pesanan):
    total = 0

    print("\n--- STRUK PEMESANAN ---")

    for i, pesanan in enumerate(daftar_pesanan, 1):
        kamar = pesanan["kamar"]
        malam = pesanan["malam"]

        print(f"{i}. ", end="")
        kamar.tampilkan_detail()

        harga = kamar.get_harga_dasar()

        if isinstance(kamar, KamarDeluxe):
            pajak = harga * 10 / 100
        else:
            pajak = harga * 5 / 100

        subtotal = kamar.hitung_harga_total(malam)

        print(f"Harga Dasar/Malam: Rp {harga:,.0f} | Pajak: Rp {pajak:,.0f}")
        print(f"Menginap: {malam} malam | Subtotal: Rp {subtotal:,.0f}")

        total += subtotal

    print("----------------------------------------")
    print(f"TOTAL TAGIHAN: Rp {total:,.0f}")
    print("----------------------------------------")


print("--- SETUP DATA KAMAR ---")

deluxe = KamarDeluxe(
    "Kamar Deluxe Sea View",
    0,
    1500000,
    "Private Pool"
)

standard = KamarStandard(
    "Kamar Standard Superior",
    0,
    500000,
    "2 Orang"
)

deluxe.tambah_stok(10)
standard.tambah_stok(-5)
standard.tambah_stok(20)

daftar_pesanan = [
    {"kamar": deluxe, "malam": 2},
    {"kamar": standard, "malam": 1}
]

proses_transaksi(daftar_pesanan)
```

**1. Abstraction**
KamarHotel digunakan sebagai class dasar dan tidak bisa dibuat menjadi objek langsung.

**2. Encapsulation**
Stok dan harga dasar dibuat private agar tidak bisa diubah sembarangan. Stok juga diberi validasi supaya tidak boleh negatif.

**3. Inheritance**
KamarDeluxe dan KamarStandard mewarisi fitur dari KamarHotel.

**4. Polymorphism**
Kedua jenis kamar memiliki method yang sama, tetapi cara menghitung pajaknya berbeda.

**5. Hasil Program**
Kamar Deluxe dipesan selama 2 malam dengan total Rp3.300.000, sedangkan Kamar Standard selama 1 malam dengan total Rp525.000. Jadi total tagihan adalah **Rp3.825.000**.
